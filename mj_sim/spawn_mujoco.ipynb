```python
import numpy as np
import mujoco as mj
import mujoco.viewer
```

```python
import sys
from pathlib import Path

_HERE = Path.cwd() if '__file__' not in globals() else Path(__file__).resolve().parent
_ROOT = _HERE.parents[0]
sys.path.insert(0, str(_HERE))

xml_path = str(_ROOT / 'models' / 'scenes' / 'g1' / 'scene_12dof_fixed.xml')

model = mujoco.MjModel.from_xml_path(xml_path)
data = mujoco.MjData(model)
```

```python
mujoco.mj_resetDataKeyframe(model, data, 0)
mujoco.mj_forward(model, data)

with mujoco.viewer.launch_passive(model, data,
                                  show_left_ui=True, show_right_ui=False) as viewer:
    while viewer.is_running():
        mujoco.mj_step(model, data)
        viewer.sync()
```
